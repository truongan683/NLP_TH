# LAB 08 - Tokenization

Chạy Run All để xem token, token IDs, special tokens và attention mask của BERT.

```python
from transformers import AutoTokenizer
from pretrained_classification import MODEL
tokenizer = AutoTokenizer.from_pretrained(MODEL)
text = 'The cats are playing.'
```

```python
encoded = tokenizer(text, return_tensors='pt')
print('keys:', list(encoded.keys()))
print('words:', text.split())
print('tokens:', tokenizer.convert_ids_to_tokens(encoded['input_ids'][0]))
print('input_ids:', encoded['input_ids'].tolist())
print('attention_mask:', encoded['attention_mask'].tolist())
```

```python
batch = tokenizer([text, 'A movie.'], padding=True, return_tensors='pt')
print(batch)
for ids, mask in zip(batch['input_ids'], batch['attention_mask']):
    tokens = tokenizer.convert_ids_to_tokens(ids)
    special = tokenizer.get_special_tokens_mask(ids.tolist(), already_has_special_tokens=True)
    print(list(zip(tokens, ids.tolist(), mask.tolist(), special)))
    print('words/subwords:', len(text.split()) if ids.equal(batch['input_ids'][0]) else 2, sum(not s for s in special))
```

`input_ids` là ID trong vocabulary. `attention_mask=1` là token hợp lệ, `0` là padding. `[CLS]` dùng ở đầu sequence, `[SEP]` kết thúc/phân tách. Số từ không nhất thiết bằng số subword token.
